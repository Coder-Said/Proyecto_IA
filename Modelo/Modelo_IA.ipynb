```python
# Configuración inicial

# Importación de librerías y configuración de variables

import os
import numpy as np
import pandas as pd
import librosa
import librosa.display
import matplotlib.pyplot as plt

RUTA_DATASET = "dataset"   # carpeta raíz, con una subcarpeta por clase
SR = 16000                 # frecuencia de muestreo común para todos los audios
EXTENSIONES = (".wav", ".mp3", ".flac", ".ogg")

# Conexión a google drive

from google.colab import drive
drive.mount('/content/drive')
RUTA_DATASET = "/content/drive/MyDrive/Car_Dataset/car diagnostics dataset"
```

```python
# Integración del dataset mediante búsqueda recursiva

registros = []

# os.walk recorre de forma recursiva todos los directorios y subdirectorios
for root, dirs, files in os.walk(RUTA_DATASET):
    for archivo in files:
        if archivo.lower().endswith(EXTENSIONES):
            ruta_completa = os.path.join(root, archivo)

            # Determinamos la etiqueta basándonos en la subcarpeta inmediata bajo RUTA_DATASET
            ruta_relativa = os.path.relpath(root, RUTA_DATASET)
            clase = ruta_relativa.split(os.sep)[0] if ruta_relativa != "." else "desconocido"

            registros.append({
                "ruta": ruta_completa,
                "etiqueta": clase
            })

df = pd.DataFrame(registros)

print(f"Total de audios encontrados: {len(df)}")
if not df.empty:
    print("\nDistribución por etiqueta:")
    print(df["etiqueta"].value_counts())

    # carga y verificación
    def cargar_audio(ruta, sr=SR):
        senal, _ = librosa.load(ruta, sr=sr, mono=True)
        return senal

    # Prueba rápida con el primer audio
    senal = cargar_audio(df.loc[0, "ruta"])
    print(f"\nPrueba rápida con el primer audio:")
    print(f"Ruta: {df.loc[0, 'ruta']}")
    print(f"Duración: {len(senal)/SR:.2f} s | Muestras: {len(senal)}")
else:
    print("No se encontraron archivos de audio. Verifica que el dataset esté subido correctamente y las extensiones coincidan.")
```
